```python
class Staff:
    def __init__(self, s_name, access_code):
        self.s_name = s_name
        self.access_code = access_code   # Uses the property setter

    # Property (Getter)
    @property
    def access_code(self):
        return "Access Denied! Unauthorized users cannot view the access code."

    # Method for authorized users
    def view_access_code(self, authorized=False):
        if authorized:
            return self.__access_code
        return "Access Denied! Unauthorized user."

    # Property Setter
    @access_code.setter
    def access_code(self, new_code):
        if len(new_code) >= 6:
            self.__access_code = new_code
            print("Access code updated successfully.")
        else:
            print("Error: Access code must be at least 6 characters long.")

    # Display staff information
    def display_info(self):
        print(f"Staff Name: {self.s_name}")
        print("Access Code: ******** (Protected)")


# Testing the program

staff1 = Staff("John Mensah", "UMAT123")

# Display staff information
staff1.display_info()

# Unauthorized access
print(staff1.access_code)

# Authorized access
print("Authorized View:", staff1.view_access_code(True))

# Update access code with valid length
staff1.access_code = "SMART789"

# View updated code
print("Updated Code:", staff1.view_access_code(True))

# Try invalid access code
staff1.access_code = "123"
```
